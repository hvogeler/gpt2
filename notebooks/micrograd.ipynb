```python
from gpt2.micrograd.micrograd import Value

a = Value(4.0, label="a")
b = Value(5.0, label="b")
c = Value(6.0, label="c")

d = a * b + c
d.label = "d"
print(f"d.prev = {d.prev}, {d.op}")
```

```python
for v in d.prev:
    print(f"v = {v.prev} - op = {v.op}")
```

```python
from gpt2.micrograd.micrograd import draw_dot

draw_dot(d)
```
