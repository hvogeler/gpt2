```python
def lol() -> None:
    a = Value(4.0, label="a")
    b = Value(5.0, label="b")
    c = Value(6.0, label="c")

    d = a * b
    d.label = "d"

    e = d + c
    e.label = "e"

    d1 = d

    h = 0.0001
    a = Value(4.0 + h, label="a")
    b = Value(5.0, label="b")
    c = Value(6.0, label="c")

    d = a * b + c
    d2 = d

    grad = (d2.data - d1.data) / h

    print(f"grad = {grad}, d1.data = {d1.data}, d2.data = {d2.data}")

lol()


```

```python
import math

import matplotlib.pyplot as plt
import numpy as np

# pyright: reportUnknownMemberType=false

def tanh(x: float) -> float:
    """Apply the tanh activation function to this Value instance."""

    t = (math.exp(2 * x) - 1) / (math.exp(2 * x) + 1)  # tanh formula
    return t

xs = np.arange(-5, 5, 0.2)
ys = [tanh(x) for x in xs]

plt.plot(xs, ys)
plt.grid(True)
plt.title("tanh(x)")
plt.xlabel("x")
plt.ylabel("tanh(x)")
plt.show()
```

```python
x1 = Value(2.0, label="x1")
x2 = Value(0.0, label="x2")

# Weights
w1 = Value(-3.0, label="w1")
w2 = Value(1.0, label="w2")

# bias
b = Value(6.8813735870195432, label="b")

# x1*w1 + x2*w2 + b
x1w1 = x1 * w1
x1w1.label = "x1w1"

x2w2 = x2 * w2
x2w2.label = "x2w2"

x1w1x2w2 = x1w1 + x2w2
x1w1x2w2.label = "x1w1x2w2"

n = x1w1x2w2 + b; n.label = "n"
o= n.tanh(); o.label = "o"
```

```python
o.grad = 1.0
o.backward()
n.backward()
x1w1x2w2.backward()
b.backward()
x1w1.backward()
x2w2.backward()
draw_dot(o)
```
